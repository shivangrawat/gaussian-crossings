We check if the dimless functions and the regular solution give the same answer.

```python
import matplotlib.pyplot as plt
import numpy as np
import torch
import scipy
import scipy.linalg
import gaussian_crossings.utils as utils
import gaussian_crossings.formula as formula
import gaussian_crossings.process as process
```

```python
# Set simulation parameters.
T = 100.0    # Total simulation time (e.g., 10 time units)
tau_e = 2.4   # Correlation decay parameter
u = 1.0
sigma = 1.0 # Standard deviation (amplitude) of the process
kappa = 0.1
```

```python
# Choose a threshold and a time interval for testing.
# Instantiate both classes.
# For the Dimless version, we pass tau=1.0 (so that 1/tau = 1).
dimless_obj = formula.GaussianUpCrossingsDimless(process.r_OU_noise, u=u, tau=tau_e, kappa=kappa, sigma=sigma)
nondim_obj  = formula.GaussianUpCrossings(process.r_OU_noise, u=u, tau=tau_e, kappa=kappa, sigma=sigma)
```

```python
print("=== Mean Rates ===")
print("Upcrossing mean rate (Dimless):", dimless_obj.upcrossing_mean_rate().item())
print("Upcrossing mean rate (Non-Dimless):", nondim_obj.upcrossing_mean_rate().item())

print("Downcrossing mean rate (Dimless):", dimless_obj.downcrossing_mean_rate().item())
print("Downcrossing mean rate (Non-Dimless):", nondim_obj.downcrossing_mean_rate().item())

print("Crossing mean rate (Dimless):", dimless_obj.crossing_mean_rate().item())
print("Crossing mean rate (Non-Dimless):", nondim_obj.crossing_mean_rate().item())
```
